## Poisson Random Variable

Given the average number phone calls recived in a minute by a person (variable `lamda_avg`), what's the probability receiving mores/less calls in the next minute (variable `x_range`)?

```python
import math

lamda_avg = 2
x_range = 4

for x in range (0, (x_range + 1)):
    val = (math.exp(-(lamda_avg)) * (lamda_avg**x)) / math.factorial(x)
    print(x, "-->", round(val, 4))
```

## Binomial discrete variable - Tossing a coin example

Given the chances of getting heads in a biased coin (variable `P_A`), what's the probability of heads in each trail (variable `x_range`)?

```python
import math

x_range = 10
P_A = 0.4

P_B = 1 - P_A
PDF = 0

for x in range (0, (x_range + 1)):
    val = (math.comb(x_range,x) * (P_A**x) * (P_B**(x_range-x)))
    PDF += val
    print(x, "-->", "P(X):", round(val, 4), "| F(X):", round(PDF, 4))
```

## Probability of a continues variable

In a given range of outcomes (between `lower_limit` and `upper_limit`), what is the possibility of getting a sub/fractional range outcome?

```python
lower_limit = -1
upper_limit = 11

sub_lower_lim = 4
sub_upper_lim = 7

probabilty_total = 1/(upper_limit - lower_limit)
probabilty_range = probabilty_total * (sub_upper_lim - sub_lower_lim)
PDF = probabilty_total * (sub_upper_lim - lower_limit)
print("P(X):", round(probabilty_range, 4), "| F(X):", round(PDF, 4))
```
